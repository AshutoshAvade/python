```python
import pandas as pd
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Generate random ages between 18 and 80
ages = np.random.randint(18, 81, size=7043).astype(float)

# Introduce null values randomly (e.g., 5% missing values)
null_indices = np.random.choice(7043, size=int(0.05 * 7043), replace=False)
ages[null_indices] = np.nan

# Create a DataFrame
age_data = pd.DataFrame(ages, columns=['Age'])

# Show the first few rows of the DataFrame
print(age_data)
```

```python
import pandas as pd
import numpy as np

# Total rows
num_rows = 7043

# Set seed for reproducibility
np.random.seed(42)

# Generate realistic ages
ages = np.random.normal(loc=40, scale=12, size=num_rows).astype(int)  # normal distribution
ages = np.clip(ages, 18, 80)  # limit values between 18 and 80

# Add ~2% missing values
num_nulls = int(num_rows * 0.02)
null_indices = np.random.choice(range(num_rows), num_nulls, replace=False)
ages = pd.Series(ages)
ages.loc[null_indices] = np.nan

# Save to CSV file (you can merge it later)
ages.to_csv('C:/Users/HP/Desktop/python/eda/age_column.csv', index=False, header=['Age'])

print("✅ Age column generated successfully with random realistic values and saved as 'age_column.csv'")
print(f"Sample values:\n{ages.head(15)}")
print(f"\nMissing values count: {ages.isna().sum()}")
```

```python
import pandas as pd
import numpy as np

# Total rows
num_rows = 7043

# Set seed for reproducibility
np.random.seed(42)

# Generate random genders
genders = np.random.choice(['Male', 'Female'], size=num_rows, p=[0.5, 0.5])

# Convert to Series
genders = pd.Series(genders)

# Add ~2% missing values
num_nulls = int(num_rows * 0.02)
null_indices = np.random.choice(genders.index, num_nulls, replace=False)
genders.loc[null_indices] = np.nan

# Save to CSV
genders.to_csv('C:/Users/HP/Desktop/python/eda/gender_column.csv', index=False, header=['Gender'])

print("✅ Gender column generated successfully and saved as 'gender_column.csv'")
print(f"Sample values:\n{genders.head(15)}")
print(f"\nMissing values count: {genders.isna().sum()}")
```
