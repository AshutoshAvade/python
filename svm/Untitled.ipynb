```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

```python
# Data
X = np.array([[-3, -5], [-2, -2], [1, 3], [2, 4]])
y = np.array([1, 1, 2, 2])
```

```python
# Train linear SVM
clf = SVC(kernel='linear')
clf.fit(X, y)
```

```python
# Predict new point
new_point = np.array([[-4, -6]])
prediction = clf.predict(new_point)
```

```python
# Get separating hyperplane parameters
w = clf.coef_[0]
b = clf.intercept_[0]
```

```python
# Create grid to plot decision boundary
x_min, x_max = -6, 6
xx = np.linspace(x_min, x_max, 100)
yy = -(w[0] * xx + b) / w[1]
```

```python
# Margin lines
margin = 1 / np.sqrt(np.sum(w ** 2))
yy_down = yy - np.sqrt(1 + (w[0]/w[1])**2) * margin
yy_up = yy + np.sqrt(1 + (w[0]/w[1])**2) * margin
```

```python
# Plot points
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], color='red', label='Class 1') 
plt.scatter(X[y == 2][:, 0], X[y == 2][:, 1], color='blue', label='Class 2')
plt.legend()
```

```python
# Plot new point
plt.scatter(new_point[0, 0], new_point[0, 1],
            color='green', marker='*', s=200,
            label=f'New point → class {prediction[0]}')
```

```python
# Plot decision boundary and margins
plt.plot(xx, yy, 'k-', label='Decision boundary')
plt.plot(xx, yy_down, 'k--')
plt.plot(xx, yy_up, 'k--')
plt.xlim(-6, 6)
plt.ylim(-7, 6)
plt.legend()
plt.title("SVM Linear Decision Boundary Visualization")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```
