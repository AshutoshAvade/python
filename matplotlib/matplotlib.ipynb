```python
import matplotlib.pyplot as plt
x=["Python","C","c++","Java"]
y=[85,70,60,82]
z=[20,30,40,50]
plt.xlabel("language",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.title("Stats")
c= ["y","b","m","g"]
plt.bar(x,y,width=0.4,color="b",label="popularity")
plt.bar(x,z,width=0.4,color="y",label="popularity1")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=["Python","C","c++","Java"]
y=[85,70,60,82]
z=[20,30,40,50]

width=0.2
p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel("language",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.title("Stats",fontsize=15)

plt.bar(p,y,width,color="b",label="popularity")
plt.bar(p1,z,width,color="y",label="popularity1")

plt.xticks(p+width/2,x,rotation=10)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
plt.scatter(day,no)
plt.title("Sdcatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
import matplotlib.pyplot as plt
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
colors=["r","y","g","b","r","g","r"]
sizes=[250,220,240,235,410,350,507]
plt.scatter(day,no,c=colors,s=sizes,alpha=0.4,marker="*",edgecolor="k",linewidth=2)
plt.title("Sdcatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
import matplotlib.pyplot as plt
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
no2=[3,2,4,5,1,6,2]
colors=[10,49,30,29,56,20,30]
sizes=[250,220,240,235,410,350,507]
plt.scatter(day,no,c=colors,s=sizes,cmap="viridis")
plt.scatter(day,no2,color="b",marker="*", s=sizes,cmap="viridis")
t=plt.colorbar()
t.set_label("Colorbar")
plt.title("Sdcatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

# stack plot

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area=[1,4,6,3,5]

plt.stackplot(x,area)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area=[1,4,6,3,5]
area1=[1,4,2,3,1]
area2=[2,3,1,4,2]

plt.stackplot(x,area,area1,area2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area=[1,4,6,3,5]
area1=[1,4,2,3,1]
area2=[2,3,1,4,2]

l=["area1","area2","area3"]

plt.stackplot(x,area,area1,area2,labels=l,colors=["y","b","c"])
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area=[1,4,6,3,5]
area1=[1,4,2,3,1]
area2=[2,3,1,4,2]

l=["area1","area2","area3"]

plt.stackplot(x,area,area1,area2,labels=l,colors=["y","b","c"],baseline="wiggle")
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area=[1,4,6,3,5]
area1=[1,4,2,3,1]
area2=[2,3,1,4,2]

l=["area1","area2","area3"]

plt.stackplot(x,area,area1,area2,labels=l,colors=["y","b","c"],baseline="zero")

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.legend(loc=2)
plt.show()
```
