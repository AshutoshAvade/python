```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
```

```python
# --- Step 1: Create a simple dataset (10 points) ---
X = np.array([
    [1, 2],
    [2, 2],
    [2, 3],
    [8, 7],
    [8, 8],
    [25, 80],
    [7, 8],
    [1, 0.6],
    [9, 7],
    [8, 6]
])
```

```python
# --- Step 2: Apply DBSCAN ---
# eps = neighborhood radius, min_samples = minimum points to form a cluster
dbscan = DBSCAN(eps=2, min_samples=2)
labels = dbscan.fit_predict(X)   
```

```python
print("Cluster Labels:", labels)
```

```python
# --- Step 3: Visualize ---
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='plasma', s=100)
plt.title("DBSCAN Clustering Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

# --- Step 1: Simple dataset (10 points) ---
X = np.array([
    [1, 2],
    [2, 2],
    [2, 3],
    [8, 7],
    [8, 8],
    [25, 80],
    [7, 8],
    [1, 0.6],
    [9, 7],
    [8, 6]
])

# --- Step 2: Apply DBSCAN ---
dbscan = DBSCAN(eps=2, min_samples=2)
labels = dbscan.fit_predict(X)
print("Cluster Labels:", labels)

# --- Step 3: Plot points ---
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='plasma', s=100, edgecolors='k')

# --- Step 4: Circle clusters manually ---
# Cluster 0 around (1–2,2–3)
circle1 = plt.Circle((1.8, 2), 1.5, color='blue', fill=False, linewidth=2, linestyle='--', label='Cluster 0')

# Cluster 1 around (8,7)
circle2 = plt.Circle((8, 7), 2, color='orange', fill=False, linewidth=2, linestyle='--', label='Cluster 1')

plt.gca().add_patch(circle1)
plt.gca().add_patch(circle2)

# --- Step 5: Labeling and visualization ---
plt.scatter(X[labels == -1, 0], X[labels == -1, 1], color='red', s=150, edgecolors='k', marker='x', label='Outlier')
plt.legend()
plt.title("DBSCAN Clustering (with Circles Around Clusters)")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```
