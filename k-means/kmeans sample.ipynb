```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Step 1: Create sample data (20 rows, 2 features)
data = {
    'Feature1': [1, 2, 1.5, 8, 9, 8.5, 1, 2.2, 1.8, 9.2,
                 5, 6, 5.5, 6.5, 5.8, 6.2, 5.4, 6.1, 5.7, 6.3],
    'Feature2': [1, 1.5, 1.2, 8, 9, 8.2, 0.8, 1.4, 1.1, 9.1,
                 5, 5.5, 5.2, 5.8, 5.6, 6.1, 5.3, 5.9, 5.7, 6]
}
```

```python
df = pd.DataFrame(data)
print("Sample Data:\n", df)
```

```python
# Step 2: Apply K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(df)
```

```python
# Step 3: Add cluster labels to the DataFrame
df['Cluster'] = kmeans.labels_
```

```python
# Step 4: Display the clustered data
print("\nClustered Data:\n", df)
```

```python
# Step 5: Visualize the clusters
plt.figure(figsize=(6, 5))
plt.scatter(df['Feature1'], df['Feature2'], c=df['Cluster'], cmap='viridis', s=80)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', s=200, alpha=0.7, marker='X', label='Centroids')
plt.title("K-Means Clustering on 20 Sample Points")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
