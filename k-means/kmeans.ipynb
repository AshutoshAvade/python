```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
# Sample data (20 points, 2 features)
X = [
    [1, 1], [1.5, 2], [2, 1.5], [8, 8], [8.5, 9],
    [9, 8.5], [9, 9], [1, 2], [2, 1], [8, 9.2],
    [5, 5], [5.5, 5.2], [6, 5.8], [6.2, 6], [5.8, 5.5],
    [6.5, 5.9], [5.3, 6], [6.1, 5.7], [5.9, 6.2], [6, 5.8]
]
```

```python
# Create and fit KMeans model with 2 clusters
kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(X)
```

```python
# Get labels for each point
labels = kmeans.labels_
```

```python
labels
```

```python
# Plot the clusters
#[x[0] for x in X]: gets all x-values (Feature 1). [x[1] for x in X]: gets all y-values (Feature 2).
plt.scatter([x[0] for x in X], [x[1] for x in X], c=labels, cmap='viridis', s=80) 
#[:, 0] gets all x-coordinates of the centers. [:, 1] → gets all y-coordinates of the centers.
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            color='red', marker='X', s=200, label='Centroids')
plt.title("K-Means Clustering (2 Clusters)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
