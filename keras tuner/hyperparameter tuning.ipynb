```python
X = df.iloc[:,:-1].values
y = df.iloc[:,-1].values 
```

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
```

```python
X = scalar.fit_transform(X)
```

```python
X
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model1 = Sequential()
model1.add(Dense(32,activation='relu',input_dim=8))
model1.add(Dense(1,activation='sigmoid'))
```

```python
model1.compile(optimizer='Adam', loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#model1.fit(X_train,y_train, batch_size=32, epochs=10)
model1.fit(X_train,y_train, batch_size=32, epochs=100,validation_data=(X_test,y_test))
```

```python
#how to automate the selection activation function using keras
import kerastuner as kt
```

```python
def build_model(hp):
    model1 = Sequential()

    model1.add(Dense(32,activation='relu',input_dim=8))
    model1.add(Dense(1,activation='sigmoid'))

    optimizer = hp.Choice('optimizer', values= ['adam','sgd','rmsprop','adadelta'])

    model1.compile(optimizer=optimizer, loss='binary_crossentropy',metrics=['accuracy'])

    return model1
```

```python
tuner = kt.RandomSearch(build_model,
                        objective='val_accuracy',
                        max_trials=5,
                        overwrite=True
)
```

```python
tuner.search(X_train,y_train, epochs=5,validation_data=(X_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model1 = tuner.get_best_models(num_models=1)[0]
```

```python
model1.summary()
```

```python
model1.fit(X_train,y_train, batch_size=32, epochs=100,initial_epoch=6, validation_data=(X_test,y_test))
```
