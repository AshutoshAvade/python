📊 Sales Data Analysis & Visualization Project
1. Project Overview

Objective:
The goal of this project is to simulate a real-world sales dataset and perform data analysis, statistical calculations, and visualizations using Python libraries:

NumPy: For generating synthetic numeric and categorical data.

Pandas: For data manipulation, cleaning, and analysis.

Matplotlib & Seaborn: For creating informative and visually appealing plots.

Key Outcomes:

Generate a realistic synthetic dataset.

Perform exploratory data analysis (EDA).

Compute statistical summaries and correlations.

Create multiple visualizations to identify trends and insights.

Derive actionable business insights from the data.

2. Dataset Description

The synthetic dataset contains 300 transactions with the following attributes:

| Column          | Description                                                 |
| --------------- | ----------------------------------------------------------- |
| `Order_ID`      | Unique identifier for each transaction                      |
| `Order_Date`    | Date of the order                                           |
| `Region`        | Geographic region of the sale (North, South, East, West)    |
| `Sales_Rep`     | Name of the sales representative                            |
| `Category`      | Product category (Electronics, Furniture, Clothing, Sports) |
| `Product`       | Specific product sold                                       |
| `Units_Sold`    | Number of units sold                                        |
| `Unit_Price`    | Price per unit of the product                               |
| `Discount`      | Discount applied on the order (percentage)                  |
| `Total_Sales`   | Revenue generated after discount                            |
| `Profit`        | Profit generated from the sale                              |
| `Profit_Margin` | Profit as a percentage of total sales                       |



Notes:

The dataset is synthetic, generated randomly using NumPy to mimic real business sales.

Relationships are created logically, e.g., profit < total sales, certain sales reps assigned to specific regions.

3. Data Analysis
3.1 Exploratory Data Analysis (EDA)

Checked dataset shape, types, and missing values.

Generated summary statistics (mean, median, std, min, max) for numeric fields.

Verified logical consistency, e.g., revenue and profit values.

3.2 Statistical Insights

Average Revenue: Shows the mean revenue across all transactions.

Maximum Profit: Highest profit generated by a transaction.

Minimum Units Sold: Lowest quantity sold.

Correlation Analysis: Identifies relationships among Units_Sold, Unit_Price, Total_Sales, and Profit.

Revenue and Profit are strongly positively correlated (~0.9).

3.3 Aggregated Insights

Revenue by Region: North and West regions lead in revenue.

Revenue by Category: Electronics generate the highest revenue.

Profit by Sales Representative: Top-performing reps identified for reward/analysis.

4. Data Visualization

The project includes the following visualizations:

| Plot                                      | Purpose                                               | Key Insight                                                           |
| ----------------------------------------- | ----------------------------------------------------- | --------------------------------------------------------------------- |
| **Histogram (Revenue)**                   | Show distribution of sales                            | Revenue is skewed; most sales are moderate                            |
| **Bar Plot (Average Profit by Category)** | Compare profit across product categories              | Electronics and Furniture are most profitable                         |
| **Scatter Plot (Units Sold vs Revenue)**  | Visualize relationship between units sold and revenue | Strong positive correlation; region-wise performance visible          |
| **Box Plot (Profit by Region)**           | Detect spread and outliers in profits                 | West region shows higher median profit                                |
| **Heatmap (Correlation)**                 | Feature correlation visualization                     | Revenue and profit highly correlated; discount reduces profit         |
| **Pair Plot**                             | Visualize pairwise relationships                      | Helps detect linear/non-linear relationships between numeric features |
| **Violin Plot (Profit by Category)**      | Distribution comparison                               | Electronics category has wider profit range                           |
| **Count Plot (Transactions by Region)**   | Frequency distribution                                | Most transactions occur in North and West                             |
| **FacetGrid (Profit by Sales Rep)**       | Multi-dimensional view (Region × Category × Profit)   | Top sales reps per category/region identified                         |
| **Regression Plot (Discount vs Revenue)** | Analyze discount impact                               | Higher discounts slightly reduce revenue                              |
