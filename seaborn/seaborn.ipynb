```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Enable inline plotting
%matplotlib inline

# Set Seaborn style
sns.set(style="whitegrid")

# ===============================================================
# Load Example Datasets
# ===============================================================
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
flights = sns.load_dataset("flights")
titanic = sns.load_dataset("titanic")

# ===============================================================
# 1️⃣ Line Plot
# ===============================================================
plt.figure(figsize=(8, 5))
sns.lineplot(x="year", y="passengers", data=flights, marker="o")
plt.title("Line Plot - Passengers Over Years", fontsize=14)
plt.show()
```

# heat map

```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Enable inline plotting
%matplotlib inline

# Set Seaborn style
sns.set(style="whitegrid")

# ===============================================================
# Load Example Datasets
# ===============================================================
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
flights = sns.load_dataset("flights")
titanic = sns.load_dataset("titanic")

flights_pivot = flights.pivot_table(
    index="month", 
    columns="year", 
    values="passengers", 
    observed=False  # Prevents FutureWarning
)

plt.figure(figsize=(10, 6))
sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt=".0f"   # ✅ Float-safe format (0 decimals)
)
plt.title("Heatmap - Flight Passengers", fontsize=14)
plt.show()
```

# facetgrid plot

```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Enable inline plotting
%matplotlib inline

# Set Seaborn style
sns.set(style="whitegrid")

# ===============================================================
# Load Example Datasets
# ===============================================================
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
flights = sns.load_dataset("flights")
titanic = sns.load_dataset("titanic")

flights_pivot = flights.pivot_table(
    index="month", 
    columns="year", 
    values="passengers", 
    observed=False  # Prevents FutureWarning
)

# ===============================================================
# 14️⃣ FacetGrid Plot
# ===============================================================
g = sns.FacetGrid(tips, col="sex", row="time", margin_titles=True, height=4)
g.map_dataframe(sns.scatterplot, x="total_bill", y="tip", hue="day")
g.add_legend()
plt.suptitle("FacetGrid - Tips by Gender and Time", y=1.02, fontsize=14)
plt.show()
```
