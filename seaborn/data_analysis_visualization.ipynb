```python
# ===============================================================
# 📊 SALES DATA ANALYSIS & VISUALIZATION PROJECT
# ===============================================================
# This notebook demonstrates:
# - Synthetic data generation using NumPy
# - Data analysis using Pandas
# - Statistical calculations
# - Visualization using Matplotlib & Seaborn
# ===============================================================

# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Enable inline plotting
%matplotlib inline
sns.set(style="whitegrid", palette="muted")

# ===============================================================
# 1️⃣ GENERATE RANDOM DATASET USING NUMPY
# ===============================================================

np.random.seed(42)

# Generate synthetic sales data
num_records = 200
data = {
    'Order_ID': np.arange(1001, 1001 + num_records),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], num_records),
    'Sales_Rep': np.random.choice(['John', 'Priya', 'Alex', 'Ravi', 'Sara', 'David'], num_records),
    'Product_Category': np.random.choice(['Electronics', 'Furniture', 'Clothing', 'Sports'], num_records),
    'Units_Sold': np.random.randint(1, 50, num_records),
    'Unit_Price': np.random.uniform(10, 500, num_records).round(2),
    'Discount': np.random.choice([0, 5, 10, 15, 20], num_records)
}

sales_df = pd.DataFrame(data)
sales_df['Total_Sales'] = (sales_df['Units_Sold'] * sales_df['Unit_Price']) * (1 - sales_df['Discount']/100)
sales_df['Profit'] = sales_df['Total_Sales'] * np.random.uniform(0.05, 0.3, num_records)

print("✅ Synthetic dataset created successfully!")
print("\nSample Records:")
print(sales_df.head())

# ===============================================================
# 2️⃣ DATA EXPLORATION AND CLEANING
# ===============================================================

print("\nDataset Info:")
print(sales_df.info())

print("\nSummary Statistics:")
print(sales_df.describe())

print("\nMissing Values per Column:")
print(sales_df.isnull().sum())

# ===============================================================
# 3️⃣ STATISTICAL ANALYSIS USING NUMPY & PANDAS
# ===============================================================

mean_sales = np.mean(sales_df['Total_Sales'])
median_sales = np.median(sales_df['Total_Sales'])
std_sales = np.std(sales_df['Total_Sales'])
max_sales = np.max(sales_df['Total_Sales'])
min_sales = np.min(sales_df['Total_Sales'])

print("\n📈 Statistical Summary:")
print(f"Mean Total Sales: {mean_sales:.2f}")
print(f"Median Total Sales: {median_sales:.2f}")
print(f"Standard Deviation: {std_sales:.2f}")
print(f"Max Sales: {max_sales:.2f}")
print(f"Min Sales: {min_sales:.2f}")

# Grouped Analysis
region_sales = sales_df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
category_sales = sales_df.groupby('Product_Category')['Total_Sales'].sum().sort_values(ascending=False)
rep_sales = sales_df.groupby('Sales_Rep')['Total_Sales'].sum().sort_values(ascending=False)

print("\n🧭 Total Sales by Region:")
print(region_sales)

print("\n🏷️ Total Sales by Product Category:")
print(category_sales)

print("\n👩‍💼 Total Sales by Sales Representative:")
print(rep_sales)

# ===============================================================
# 4️⃣ DATA VISUALIZATION USING MATPLOTLIB & SEABORN
# ===============================================================

# 1. Histogram of Total Sales
plt.figure(figsize=(8,5))
sns.histplot(sales_df['Total_Sales'], bins=25, kde=True, color='skyblue')
plt.title("Distribution of Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.show()

# 2. Bar Plot - Total Sales by Region
plt.figure(figsize=(8,5))
sns.barplot(x=region_sales.index, y=region_sales.values, palette="coolwarm")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()

# 3. Box Plot - Sales by Product Category
plt.figure(figsize=(8,5))
sns.boxplot(x='Product_Category', y='Total_Sales', data=sales_df, palette="mako")
plt.title("Total Sales Distribution by Product Category")
plt.show()

# 4. Scatter Plot - Units Sold vs Total Sales
plt.figure(figsize=(8,5))
sns.scatterplot(x='Units_Sold', y='Total_Sales', hue='Region', data=sales_df, palette='husl')
plt.title("Units Sold vs Total Sales")
plt.show()

# 5. Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(sales_df[['Units_Sold','Unit_Price','Discount','Total_Sales','Profit']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# 6. Count Plot - Sales by Representative
plt.figure(figsize=(8,5))
sns.countplot(x='Sales_Rep', data=sales_df, palette='pastel', order=sales_df['Sales_Rep'].value_counts().index)
plt.title("Number of Orders per Sales Representative")
plt.show()

# 7. Violin Plot - Profit by Region
plt.figure(figsize=(8,5))
sns.violinplot(x='Region', y='Profit', data=sales_df, palette='Set2')
plt.title("Profit Distribution by Region")
plt.show()

# 8. Pair Plot
sns.pairplot(sales_df[['Units_Sold','Unit_Price','Discount','Total_Sales','Profit']])
plt.suptitle("Pairwise Relationships in Sales Data", y=1.02)
plt.show()


```

```python
# ===============================================================
# 📊 ENHANCED SALES DATA ANALYSIS & VISUALIZATION PROJECT
# ===============================================================
# Includes: realistic dataset, advanced analytics, and professional visualizations
# ===============================================================

# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

# Enable inline plotting
%matplotlib inline
sns.set(style="whitegrid", palette="deep")

# ===============================================================
# 1️⃣ DATA GENERATION (REALISTIC SYNTHETIC DATA)
# ===============================================================

np.random.seed(42)
num_records = 300

regions = ['North', 'South', 'East', 'West']
reps_by_region = {
    'North': ['John', 'David', 'Amar'],
    'South': ['Priya', 'Ravi', 'Sneha'],
    'East': ['Alex', 'Sara', 'Rahul'],
    'West': ['Isha', 'Karan', 'Rita']
}

products = ['Laptop', 'Chair', 'T-Shirt', 'Cricket Bat', 'Headphones', 'Table', 'Shoes', 'TV']
categories = {
    'Laptop': 'Electronics', 'TV': 'Electronics', 'Headphones': 'Electronics',
    'Chair': 'Furniture', 'Table': 'Furniture',
    'T-Shirt': 'Clothing', 'Shoes': 'Clothing',
    'Cricket Bat': 'Sports'
}

start_date = datetime(2024, 1, 1)
dates = [start_date + timedelta(days=int(x)) for x in np.random.randint(0, 365, num_records)]

data = []
for i in range(num_records):
    region = np.random.choice(regions)
    rep = np.random.choice(reps_by_region[region])
    product = np.random.choice(products)
    category = categories[product]
    units = np.random.randint(1, 50)
    price = round(np.random.uniform(20, 1000), 2)
    discount = np.random.choice([0, 5, 10, 15, 20])
    total_sales = (units * price) * (1 - discount/100)
    profit = total_sales * np.random.uniform(0.05, 0.35)
    data.append([1001+i, dates[i], region, rep, category, product, units, price, discount, total_sales, profit])

sales_df = pd.DataFrame(data, columns=[
    'Order_ID','Order_Date','Region','Sales_Rep','Category','Product','Units_Sold',
    'Unit_Price','Discount','Total_Sales','Profit'
])

sales_df['Month'] = pd.to_datetime(sales_df['Order_Date']).dt.strftime('%b')
sales_df['Profit_Margin'] = (sales_df['Profit'] / sales_df['Total_Sales'] * 100).round(2)

print("✅ Enhanced synthetic dataset created successfully!")
display(sales_df.head())

# ===============================================================
# 2️⃣ DATA OVERVIEW & SUMMARY
# ===============================================================

print("\nBasic Info:")
sales_df.info()

print("\nDescriptive Statistics:")
display(sales_df.describe())

print("\nMissing Values:")
print(sales_df.isnull().sum())

# KPIs
print("\n📊 KEY KPIs:")
print(f"Total Revenue: ₹{sales_df['Total_Sales'].sum():,.2f}")
print(f"Total Profit: ₹{sales_df['Profit'].sum():,.2f}")
print(f"Average Profit Margin: {sales_df['Profit_Margin'].mean():.2f}%")
print(f"Total Orders: {len(sales_df)}")

# ===============================================================
# 3️⃣ STATISTICAL INSIGHTS
# ===============================================================

region_sales = sales_df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
category_sales = sales_df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
rep_performance = sales_df.groupby('Sales_Rep')['Profit'].sum().sort_values(ascending=False)
month_trend = sales_df.groupby('Month')['Total_Sales'].sum().reindex(
    ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])

# ===============================================================
# 4️⃣ VISUALIZATIONS
# ===============================================================

# 1. Monthly Sales Trend
plt.figure(figsize=(10,5))
sns.lineplot(x=month_trend.index, y=month_trend.values, marker="o", color="teal")
plt.title("📆 Monthly Sales Trend")
plt.ylabel("Total Sales (₹)")
plt.show()

# 2. Sales by Region (Bar Chart)
plt.figure(figsize=(8,5))
sns.barplot(x=region_sales.index, y=region_sales.values, palette="coolwarm")
plt.title("🧭 Total Sales by Region")
plt.ylabel("Total Sales (₹)")
plt.show()

# 3. Pie Chart - Category Share
plt.figure(figsize=(6,6))
plt.pie(category_sales, labels=category_sales.index, autopct='%1.1f%%', colors=sns.color_palette("Set3"))
plt.title("📊 Sales Share by Product Category")
plt.show()

# 4. Regression Plot - Discount vs Total Sales
plt.figure(figsize=(8,5))
sns.regplot(x='Discount', y='Total_Sales', data=sales_df, scatter_kws={'alpha':0.6}, color="purple")
plt.title("💸 Discount vs Total Sales")
plt.show()

# 5. Violin Plot - Profit by Category
plt.figure(figsize=(9,5))
sns.violinplot(x='Category', y='Profit', data=sales_df, palette="mako")
plt.title("Profit Distribution by Category")
plt.show()

# 6. Top 10 Products by Revenue
top_products = sales_df.groupby('Product')['Total_Sales'].sum().nlargest(10)
plt.figure(figsize=(9,5))
sns.barplot(x=top_products.values, y=top_products.index, palette="crest")
plt.title("🏆 Top 10 Products by Revenue")
plt.xlabel("Total Sales (₹)")
plt.show()

# 7. Heatmap - Correlation
plt.figure(figsize=(8,6))
sns.heatmap(sales_df[['Units_Sold','Unit_Price','Discount','Total_Sales','Profit','Profit_Margin']].corr(),
            annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()

# 8. Swarm Plot - Profit Margin by Region
plt.figure(figsize=(9,5))
sns.swarmplot(x='Region', y='Profit_Margin', data=sales_df, palette="Set2", size=5)
plt.title("Profit Margin Distribution by Region")
plt.show()

# 9. FacetGrid - Region vs Category vs Profit
g = sns.FacetGrid(sales_df, col="Region", row="Category", height=3.5)
g.map_dataframe(sns.barplot, x="Sales_Rep", y="Profit", hue="Sales_Rep", errorbar=None, palette="muted", legend=False)
g.add_legend()
plt.subplots_adjust(top=0.92)
g.fig.suptitle("📈 Profit by Sales Rep across Region and Category")
plt.show()


# ===============================================================
# 5️⃣ INSIGHTS SUMMARY
# ===============================================================
print("\n🔍 KEY INSIGHTS:")
print(f"- '{region_sales.idxmax()}' region leads in total revenue.")
print(f"- '{category_sales.idxmax()}' is the highest-grossing category.")
print(f"- '{rep_performance.idxmax()}' is the top-performing sales representative.")
print(f"- Discounts negatively correlate with profit margin (see heatmap).")
print(f"- Sales show a steady monthly increase, peaking around {month_trend.idxmax()}.")

print("\n✅ Enhanced Data Analysis & Visualization Completed Successfully!")
```
