```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer=load_breast_cancer()
```

```python
cancer.keys()
```

```python
df=pd.DataFrame(cancer['data'],columns=cancer['feature_names'])
```

```python
df.head(5)
```

```python
from sklearn.preprocessing import MinMaxScaler # coonverts the data in range of 0-1
from sklearn.preprocessing import StandardScaler 
```

```python
scaler=StandardScaler() #sd=1 and mean=0
scaler.fit(df)
```

```python
scaled_data=scaler.transform(df)
```

```python
scaled_data
```

```python
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=2) # no of components we want to reduce the dimensions to. i.e we have 30 dimensions here it will get converted to 2
```

```python
pca.fit(scaled_data)
```

```python
x_pca=pca.transform(scaled_data)
```

```python
scaled_data.shape
```

```python
x_pca.shape
```

```python
scaled_data
```

```python
x_pca
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:,0],x_pca[:,1],c=cancer['target'])
plt.xlabel('First principle component')
plt.ylabel('Second principle component')
```
