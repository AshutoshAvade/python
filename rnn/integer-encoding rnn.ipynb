```python
import numpy as np

docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer # Tokenizer converts captial letter to small,separate it word by word,remove special symbol
tokenizer = Tokenizer(oov_token='<nothing>') # oov token is out of vocabulary token [can give any name],new word replace by oov token in training data
```

```python
tokenizer.fit_on_texts(docs) #pass the doc and give index the work that tokenizer does
```

```python
tokenizer.word_index #give the no. of uniquie words and thier index
```

```python
tokenizer.word_counts #give the frequency or how many times the word is repeated
```

```python
tokenizer.document_count #no, of rows
```

```python
sequences = tokenizer.texts_to_sequences(docs) #generates sequence for every sentence
sequences
```

```python
from keras.utils import pad_sequences 
```

```python
sequences = pad_sequences(sequences,padding='post') #gives padding to give equal size. the post means give padding afterwards
```

```python
sequences #ready input data
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
len(X_train[2])
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50) #trim to 50words to reduce time
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(SimpleRNN(32,input_shape=(50,1),return_sequences=False)) #50 is time step. [1,32 weights],32bias,32×32weights=1088 and 32+1=33bias. return_sequences=normal flow,if true then it take the out of the previous as well 
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```
