### for 2D

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Create a simple 2D dataset
X = np.array([
    [1, 2],
    [2, 1],
    [6, 8],
    [7, 7],
    [8, 6]
])

# Class
y = np.array([0, 0, 1, 1, 1])  # 0 = Class A, 1 = Class B
```

```python
# Train KNN model
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)
```

```python
# New data point
new_point = np.array([[4, 4]])
```

```python
# Predict
prediction = knn.predict(new_point)
print("Predicted class:", prediction)
```

```python
# Plot class 0 points (red)
plt.scatter(X[y==0, 0], X[y==0, 1], color='red', label='Class 0', s=80)

# Plot class 1 points (blue)
plt.scatter(X[y==1, 0], X[y==1, 1], color='blue', label='Class 1', s=80)

# Plot new point (green star)
plt.scatter(new_point[:,0], new_point[:,1], color='green', marker='*', s=200, label='New Point')
```

```python
# Plot new point (green star)
plt.scatter(new_point[:,0], new_point[:,1], color='green', marker='*', s=200, label='New Point')

# Add labels and title
plt.title('KNN Classification (k=3)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```
