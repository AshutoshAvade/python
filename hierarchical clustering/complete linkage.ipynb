```python
# ---- COMPLETE LINKAGE EXAMPLE ----
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering
```

```python
# Step 1: Same small dataset (10 samples, 2 features)
data = pd.DataFrame({
    'X': [1, 2, 3, 5, 8, 10, 12, 20, 22, 25],
    'Y': [1, 1.5, 2, 4, 8, 10, 11, 21, 22, 25]
})
```

```python
print("Dataset:")
print(data)
```

```python
# Step 2: Plot original data
plt.scatter(data['X'], data['Y'], color='blue')
plt.title("Original Data Points")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Step 3: Dendrogram for Complete Linkage
linked_complete = linkage(data, method='complete')
```

```python
plt.figure(figsize=(7,4))
dendrogram(linked_complete,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.title("Dendrogram - Complete Linkage")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
# Step 4: Agglomerative Clustering (Complete Linkage)
cluster_complete = AgglomerativeClustering(n_clusters=3, linkage='complete', metric='euclidean')
data['Cluster_Complete'] = cluster_complete.fit_predict(data)
```

```python
# Step 5: Visualize clusters
plt.scatter(data['X'], data['Y'], c=data['Cluster_Complete'], cmap='rainbow')
plt.title("Clusters Formed using Complete Linkage")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Step 6: Display cluster assignments
print("\nClustered Data (Complete Linkage):")
print(data)
```
